```python
#  Author: <Syed Hussain>
# Assignment: #1

# Is a string (str)
gym_member = "Alex Alliton"

# Is a float (float)
preferred_weight_kg = 20.5

# Is an integer (int)
highest_reps = 25

#Is a boolean (bool)
membership_active = True

# Dictionary with keys as strings and values with tuples containing three integers
workout_stats = {
    "Alex": (30, 45, 20),
    "Jamie": (25, 35, 25),
    "Taylor": (35, 30, 30)
}

# Calculating total workout minutes between friends and addition of new key pairs
for friend in workout_stats.copy():
    total_minutes = sum(workout_stats[friend])
    workout_stats [f"{friend}_Total"] = total_minutes

# 2D nested list that extracts workout minutes per friend
workout_list = [list(workout_stats[friend]) for friend in workout_stats.keys()]
print(workout_list)

yoga_running = [row[0:2] for row in workout_list]
print ("Yoga running minutes for all friends:")
print (yoga_running)

weightlifting_last_two = [row[2:] for row in workout_list[-2]]
print ("Weightlifting minutes for last two friends:")
print (weightlifting_last_two)

if friend in workout_stats:
    if friend.endswith("_Total"):
        total_minutes = sum(workout_stats[friend])

        if total_minutes >= 120:
            print (f"Good job staying active, {friend}!")

friend_name = input("Enter a friend's name: ")

if friend_name in workout_stats:
    yoga, running, weightlifting = workout_stats[friend_name]
    total_minutes = workout_stats[f"{friend_name}_Total"]
    print(f"Workout stats for {friend_name}:")
    print(f"Yoga: {yoga} minutes")
    print(f"Running: {running} minutes")
    print(f"Weightlifting: {weightlifting} minutes")
    print(f"Total minutes: {total_minutes} minutes")
else:
    print(f"Sorry, {friend_name} doesn't exist.")\

max_friend = max(workout_stats, key=workout_stats.get)
min_friend = min(workout_stats, key=workout_stats.get)

max_friend_name = max_friend.replace("_Total", "")
print(f"{max_friend_name} has the highest workout minutes with {workout_stats[max_friend]} minutes.")

min_friend_name = min_friend.replace("_Total", "")
print(f"{min_friend_name} has the lowest workout minutes with {workout_stats[min_friend]} minutes.")
```
